```python
import numpy as np
import pandas as pd

# Read the dataset
data = pd.read_csv("college friends.csv")

# Get sender names
senders = data["sender"].values

# Get messages
messages = data["message"].values

# Get timestamps
timestamps = pd.to_datetime(data["timestamp"])


# 1. Basic Information
total_messages = len(messages)

unique_members = np.unique(senders)
total_members = len(unique_members)

print("\n___________________𝔾ℝ𝕆𝕌ℙ𝔻ℕ𝔸___________________")

print("Total Members  :", total_members)
print("Total Messages :", total_messages)

# 2. Messages by Each Member
member_counts = []

for member in unique_members:
    count = np.sum(senders == member)
    member_counts.append(count)

# Convert to NumPy array
member_counts = np.array(member_counts)

print("\n______________𝕄𝔼𝕄𝔹𝔼ℝ 𝔸ℂ𝕋𝕀𝕍𝕀𝕋𝕐__________________")

for i in range(total_members):
    print(unique_members[i], ":", member_counts[i], "messages")

# 3. Most Active Member
most_active_index = np.argmax(member_counts)

most_active_member = unique_members[most_active_index]
most_active_count = member_counts[most_active_index]

print("\n____________𝕄𝕆𝕊𝕋 𝔸ℂ𝕋𝕀𝕍𝔼 𝕄𝔼𝕄𝔹𝔼ℝ________________")
print(most_active_member)
print("Messages:", most_active_count)

# 4. Least Active Member
least_active_index = np.argmin(member_counts)

least_active_member = unique_members[least_active_index]
least_active_count = member_counts[least_active_index]

print("\n_____________𝕃𝔼𝔸𝕊𝕋 𝔸ℂ𝕋𝕀𝕍𝔼 𝕄𝔼𝕄𝔹𝔼ℝ_________________")
print(least_active_member)
print("Messages:", least_active_count)


# 5. Average Messages
average_messages = np.mean(member_counts)

print("\n______________𝔸𝕍𝔼ℝ𝔸𝔾𝔼 𝕄𝔼𝕊𝕊𝔸𝔾𝔼𝕊 ℙ𝔼ℝ 𝕄𝔼𝕄𝔹𝔼ℝ_____________")
print(round(average_messages, 2))


# 6. Most Active Hour
hours = timestamps.dt.hour.values

hour_counts = np.bincount(hours, minlength=24)

most_active_hour = np.argmax(hour_counts)

print("\n_____________𝕄𝕆𝕊𝕋 𝔸ℂ𝕋𝕀𝕍𝔼 ℍ𝕆𝕌ℝ_________________")
print(most_active_hour, ":00")
print("Messages:", hour_counts[most_active_hour])


# 7. Top 5 Members
sorted_indexes = np.argsort(member_counts)[::-1]

print("\n_______________𝕋𝕆ℙ 𝟝 ℂℍ𝔸𝕋𝕋𝔼ℝ𝕊__________________")

for i in range(5):
    index = sorted_indexes[i]

    print(
        i + 1,
        ".",
        unique_members[index],
        "-",
        member_counts[index],
        "messages"
    )


# 8. Group Activity Statistics
print("\n_______________𝔾ℝ𝕆𝕌ℙ 𝕊𝕋𝔸𝕋𝕀𝕊𝕋𝕀ℂ𝕊_________________")

print("Maximum messages :", np.max(member_counts))
print("Minimum messages :", np.min(member_counts))
print("Average messages :", round(np.mean(member_counts), 2))
print("Total messages   :", np.sum(member_counts))
print("\n__*__*__*__*𝔾ℝ𝕆𝕌ℙ𝔻ℕ𝔸 ℂ𝕆𝕄ℙ𝕃𝔼𝕋𝔼*__*__*__*__")
```
